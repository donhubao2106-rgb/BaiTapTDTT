### Bài 1:Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi. Chú ý: kiểm tra chỉ xử lý các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý.

```python
n = int(input())

if n >= 100 and n <= 999:
    a = n // 100
    b = (n // 10) % 10
    c = n % 10
    dao_nguoc = c * 100 + b * 10 + a
    print("Số đảo ngược:", dao_nguoc)
else:
    print("Số không hợp lệ")
```

### Bài 3: Kiểm tra xem một số có phải là lũy thừa của 2 hay không Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không. 

```python
n = int(input())

if n > 0 and (n & (n - 1)) == 0:
    print(n, "là lũy thừa của 2")
else:
    print(n, "không phải lũy thừa của 2")
```

### Bài 5: Viêt chương trình chia một số m cho n, kết quả thu được làm tròn lên 

```python
m = int(input())
n = int(input())

if n != 0:
    kq = (m + n - 1) // n
    print(kq)
else:
    print("Không thể chia cho 0")
```
