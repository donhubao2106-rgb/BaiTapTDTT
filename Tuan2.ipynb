### Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là độ dài ba cạnh hay không, nếu  không thì in  ra màn  hình  'Khong la ba  canhcua mot tam  giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn hình. 

```python
a = float(input())
b = float(input())
c = float(input())

if a + b <= c or a + c <= b or b + c <= a:
    print("Khong la ba canh cua mot tam giac")
else:
    chu_vi = a + b + c
    s = chu_vi / 2
    dien_tich = (s * (s - a) * (s - b) * (s - c)) ** 0.5

    if a == b == c:
        print("Tam giác đều")
    elif a == b or b == c or a == c:
        if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
            print("Tam giác vuông cân")
        else:
            print("Tam giác cân")
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        print("Tam giác vuông")
    else:
        print("Tam giác thường")

    print("Chu vi:", chu_vi)
    print("Diện tích:", dien_tich)
```

### Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0) 

```python
a = float(input())
b = float(input())
c = float(input())

if a == 0:
    if b == 0:
        if c == 0:
            print("Phương trình vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        print("x =", -c / b)
else:
    delta = b**2 - 4*a*c
    if delta > 0:
        x1 = (-b + delta**0.5) / (2*a)
        x2 = (-b - delta**0.5) / (2*a)
        print("x1 =", x1, ", x2 =", x2)
    elif delta == 0:
        print("x =", -b / (2*a))
    else:
        print("Phương trình vô nghiệm")
```

### Bài 6: Tìm giá trị lớn nhất trong 4 số a, b, c, d (a, b, c, được nhập từ bàn phím).
