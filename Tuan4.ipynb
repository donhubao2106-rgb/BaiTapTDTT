```python
# Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.
n = int(input())
buoc = 0

while True:
    tam = n
    rev = 0
    while tam > 0:
        rev = rev * 10 + tam % 10
        tam //= 10
    
    if n == rev:
        break
    
    n += rev
    buoc += 1

print(f"Số bước: {buoc}")
print(f"Palindrome: {n}")
```

```python
# Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n." 
n = int(input())
if n >= 2:
    max = 2
    for i in range(2, int(n ** 0.5 + 1)):
        if n % i == 0:
            max = i
        while n % i == 0:
            n //= i
        if n > 1:
            max = n
print(max)
```

```python
# VIết chương trình nhập vào một số tới khi được một số nguyên dương thì thôi.Kiểm tra số này có phải là số nguyên tố hay không?
while True:
    n = int(input())
    if n > 0:
        break

is_prime = True
if n == 1:
    is_prime = False
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{n} là số nguyên tố")
else:
    print(f"{n} không phải là số nguyên tố")
```

```python
 #Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn [a, b]. Mỗi dòng là một bộ test (a b)."
a, b = map(int, input().split())
if a <= b:
    prime_sum = 0
    for num in range(a, b + 1):
        if num > 1:
            is_prime = True
            for i in range(2, int(num ** 0.5) + 1):
                if num % i == 0:
                    is_prime = False
                    break
            if is_prime:
                prime_sum += num
    print(prime_sum)
```

```python
#Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không
while True:
    n = int(input())
    for k in range(0, n + 1):
        if 2 ** k == n:
            print("Juan roi em")
            break
        if 2 ** k > n:
            print("Chua juan dau em")
            break
```

```python
#Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ
n = int(input())
chan = 0
le = 0
while n > 0:
    if (n % 10) % 2 == 0:
        chan += 1
    if (n % 10) % 2 != 0:
        le += 1
    n //= 10
print(chan)
print(le)
```
